```python
import pandas as pd

# Load the test dataset
df = pd.read_csv('/content/cleaned_test_data.csv')
print(df.head())
```

```python
#Define Features (X)
x = df.drop(columns=['Customer_ID'])
```

```python
import joblib

# Load joblib files
# Best model - LR
logreg = joblib.load('logreg.joblib')

# Load the saved encoder
le = joblib.load('/content/LabelEncoder.joblib')

label_mapping = {label: idx for idx, label in enumerate(le.classes_)}

# Load the scaler
scaler = joblib.load("/content/scaler.joblib")
```

```python
from sklearn.preprocessing import LabelEncoder
import numpy as np

# Label encoding for outlet city
# Get a list of known cities from the LabelEncoder
known_cities = list(le.classes_)

# Update the LabelEncoder with new cities if found in the test data
new_cities = list(set(x['outlet_city']) - set(known_cities))
if new_cities:
    le.classes_ = np.concatenate((le.classes_, new_cities))

# Perform the transformation
x['outlet_city'] = le.transform(x['outlet_city'])

# Map using saved mapping, assign -1 to unknowns
x['outlet_city'] = x['outlet_city'].map(label_mapping).fillna(-1).astype(int)
```

```python
from sklearn. preprocessing import RobustScaler

#fit &transform training data, transform testing data
x_scaled = scaler.transform(x)
```

```python
predictions = logreg.predict(x_scaled)
```

```python
Predicted_test_df = pd.DataFrame({ "Customer_ID": df["Customer_ID"], # Changed 'test_data' to 'df'
                                   "Cluster_Category": predictions })  # predicted categories

# Save to CSV
Predicted_test_df.to_csv("Predicted_test.csv", index=False)

```
